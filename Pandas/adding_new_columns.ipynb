```python
import pandas as pd

df = pd.read_csv('student_data.csv')
df.head(3)
```

Add a new column with same value for all rows

```python
df['Country'] = 'Bangladesh'
df.head(5)
```

New column with Total Marks

```python
df['Total Marks'] = df['Data Structure Marks'] + df['Python Marks'] + df['Algorithm Marks']
df.head(3)
```

New column 'Username'

```python
df['Username'] = df['FullName'].str.split(' ').str[0].str.lower()
df.sample(4)

# drop a column
# df.drop('Username',axis=1)
```

Save the updated dataframe as csv file

```python
df.to_csv('New Files/student_data_updated.csv',index=False)
```
