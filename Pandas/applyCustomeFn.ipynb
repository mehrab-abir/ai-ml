```python
import pandas as pd

df = pd.read_csv('student_data.csv')
df.sample(3)
```

Apply a custome function to any column - apply()

```python
df['Total Marks'] = df.iloc[:,2:5].sum(axis=1)

maxMark = df['Total Marks'].max()
minMark = df['Total Marks'].min()

df['Scaled Mark'] = df['Total Marks'].apply(lambda x:(x-minMark)/(maxMark-minMark))
df.sample(6)
```

```python
# seperate function
def assign_grade(mark):
    if(mark >= 260):
        return 'A+'
    elif(mark >= 200):
        return 'A'
    else:
        return 'B'
    
df["Grade"] = df['Total Marks'].apply(assign_grade)
# df['Total Marks'].apply(assign_grade) will take one value of df['Total Marks'] in each iteration and execute the assign_grade function
df.sample(5)
```
