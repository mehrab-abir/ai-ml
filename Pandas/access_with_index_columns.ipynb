### Index and column changing and iloc

index change

```python
# id_index = df.set_index('StudentID', inplace=True) # inplace=True will change the name in the original df table
id_index = df.set_index('StudentID')

# df  # -> normal 0,1,2 as index

# id_index # -> now studentId is used as index
```

iloc -- access with updated index

```python
# id_index.loc[3:5]  # won't work, because numerical index doesn't exist in the id_index
# use iloc where index is not numerical

# table_name[start_index : end_index(excluded)]
# id_index.iloc[3:5] # index 3 to 4 (not 5)
id_index.iloc[3:6] # index 3 to 5 (not 6)
```

Change column name

```python
df.rename(columns={'FullName': 'Full Name','Data Structure Marks': 'DS Marks'},inplace=True) # inplace=True changes the column name in the original df table

df
```
