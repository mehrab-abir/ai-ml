### From 2D array

```python
import pandas as pd

list_2d = [['Jhon Doe',25],['Chris Evans',40],['Robert Jr.',[50]]]

list_2d_df = pd.DataFrame(list_2d)

print(list_2d_df)
```

With custom column name

```python
import pandas as pd

list_2d = [['Jhon Doe',25],['Chris Evans',40],['Robert Jr.',[50]]] 

# pd.DataFrame(list_2d,columns=['Name','Age'])
print((pd.DataFrame(list_2d,columns=['Name','Age'])))
```

With Custom index/row name

```python
import pandas as pd

list_2d = [['Jhon Doe',25],['Chris Evans',40],['Robert Jr.',[50]]] 

# pd.DataFrame(list_2d,columns=['Name','Age'])
print((pd.DataFrame(list_2d,columns=['Name','Age'], index=[1,2,3])))
```

### From tuple to df

```python
tuple_st = (('Jhon Doe',20),('Alice Kim',65))

print(pd.DataFrame(tuple_st, columns=['Name','Age']))
```
