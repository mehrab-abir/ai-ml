```python
import pandas as pd

data = {
    "Year": [2010, 2012, 2014, 2015, 2016, 2018, 2020, 2022, 2024, 2026],
    "Programming": [761, 795, 820, 810, 850, 900, 880, 920, 950, 990],
    "Digital Marketing": [526, 556, 580, 600, 590, 620, 600, 680, 700, 650]
}

df = pd.DataFrame(data)
df
```

Legend()

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Number of Enrollment")
plt.title("Programming and Digital Marketing enrollment over the years")

plt.plot(df['Year'],df['Programming'],label="Programming")
plt.plot(df['Year'],df['Digital Marketing'],label="Digital Marketing")

plt.legend() # by default- location is the best suitable one
# plt.legend(loc='lower right')
# plt.legend(loc='upper right')
# plt.legend(loc='lower left')
```

Stylizing the lines

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Number of Enrollment")
plt.title("Programming and Digital Marketing enrollment over the years")

plt.plot(df['Year'],df['Programming'],label="Programming",color="blue")
plt.plot(df['Year'],df['Digital Marketing'],label="Digital Marketing",color="purple")

plt.legend()
```

Line width

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Number of Enrollment")
plt.title("Programming and Digital Marketing enrollment over the years")

plt.plot(df['Year'],df['Programming'],label="Programming",color="blue",linewidth=3)
plt.plot(df['Year'],df['Digital Marketing'],label="Digital Marketing",color="purple",linewidth=3)

plt.legend()
```

Line style --solid, dashed, dot etc.

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Number of Enrollment")
plt.title("Programming and Digital Marketing enrollment over the years")

plt.plot(df['Year'],df['Programming'],label="Programming",color="blue",linewidth=2, linestyle="solid")
plt.plot(df['Year'],df['Digital Marketing'],label="Digital Marketing",color="purple",linewidth=2, linestyle="dotted")
# linestyle="dashdot/solid/dotted/dash"

plt.legend()
```

Marker - the location of the points(coordinate)<br>Grid

```python
import matplotlib.pyplot as plt

plt.xlabel("Year")
plt.ylabel("Number of Enrollment")
plt.title("Programming and Digital Marketing enrollment over the years")

plt.plot(df['Year'],df['Programming'],label="Programming",color="#00804F",marker='o')
plt.plot(df['Year'],df['Digital Marketing'],label="Digital Marketing",color="#A502E0",marker='*',markersize=10)

plt.legend()
plt.grid()
```
