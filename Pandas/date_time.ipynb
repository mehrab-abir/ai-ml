```python
import pandas as pd

df = pd.read_csv('student_completed_data.csv')

df.sample(5)
```

```python
df['EnrollmentDate'] # see the data type
```

```python
df['EnrollmentDate'] = pd.to_datetime(df['EnrollmentDate'])
df['EnrollmentDate'] # see the data type now - datetime
```

```python
df['Enrollment Year'] = df['EnrollmentDate'].dt.year
df.sample(5)
```

```python
df['FinishedDate'] = pd.to_datetime(df['FinishedDate'])
```

```python
df['Duration'] = df['FinishedDate'] - df['EnrollmentDate']
df.sample(5)
```
