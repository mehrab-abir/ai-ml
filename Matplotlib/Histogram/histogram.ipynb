```python
import pandas as pd

df = pd.read_csv('model_data.csv')
df.sample(5)
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Social Media Hours")
plt.ylabel("Number of Students")
plt.title("Number of students in range of social media usage")

plt.hist(df['social_media_hours'],bins=12,edgecolor='black')
#change the number of bins and see the output
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Social Media Hours")
plt.ylabel("Number of Students")
plt.title("Number of students in range of social media usage")

plt.hist(df['social_media_hours'],bins=[2,4,6,8,10,12,14,16],edgecolor='black',color="orange")
#change the number of bins and see the output
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Study Hours")
plt.ylabel("Number of Students")
plt.title("Number of students in range of study hours")

plt.hist(df['study_hours'],bins=14,edgecolor='black',color="blue")
#change the number of bins and see the output

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Exam score")
plt.ylabel("Number of Students")
plt.title("Number of students in range of exam score")

plt.hist(df['exam_score'],bins=20,edgecolor='white',color="purple")
plt.show()
```
