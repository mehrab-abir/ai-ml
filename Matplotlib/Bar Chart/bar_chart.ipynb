```python
import pandas as pd

df = pd.read_csv("student_data.csv")
df.sample(3)
```

```python
location_group = df.groupby('Location').size()
print(location_group)
print(location_group.index)
print(location_group.values)
```

```python
import matplotlib.pyplot as plt

plt.bar(location_group.index, location_group.values,color="#007185")
```

```python
status_group = df.groupby('CompletionStatus').size()
plt.bar(status_group.index,status_group.values,color=['green','purple','red'])

plt.xlabel("Completion Status")
plt.ylabel("Number of students")
plt.title("Number of students by completion status")
```
