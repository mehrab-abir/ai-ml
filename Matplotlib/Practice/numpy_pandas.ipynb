```python
import pandas as pd

df = pd.read_csv("Practice.csv")
df
```

```python
df['Discounted_Price'] = df['Price'] - (df['Price'] * 0.1)
df.to_csv("discounted_products.csv")
```

Check duplicate and clean

```python
df.duplicated(subset='EmployeeID').sum()
cleaned_df = df.drop_duplicates(subset='EmployeeID')
cleaned_df
```

Handling missing values

```python
count_missing_age = df['Age'].isnull().sum()
count_missing_salary = df['Salary'].isnull().sum()

print(count_missing_age,count_missing_salary)

cleaned_df['Age'] = cleaned_df['Age'].fillna(cleaned_df['Age'].mean(),inplace=True)
cleaned_df
```

```python
cleaned_df.dropna(subset=['Salary'],inplace=True)
```

Statistical Functions

```python
marks1_min = cleaned_df['Marks1'].min()
marks2_min = cleaned_df['Marks2'].min()
marks3_min = cleaned_df['Marks3'].min()

print(f"Min of\nMarks1: {marks1_min}\nMarks2: {marks2_min}\nMarks3: {marks3_min}")

marks1_mean = cleaned_df['Marks1'].mean()
marks2_mean = cleaned_df['Marks2'].mean()
marks3_mean = cleaned_df['Marks3'].mean()

print(f"\nMean of\nMarks1: {marks1_mean}\nMarks2: {marks2_mean}\nMarks3: {marks3_mean}")
```

```python
cleaned_df['Total Marks'] = cleaned_df.iloc[:,5:8].sum(axis=1)
cleaned_df
```

```python
cleaned_df['Total Marks'].std()

cleaned_df.to_csv('cleaned_data.csv')
```
