```python
import pandas as pd

df = pd.read_csv('cleaned_data.csv')
df
```

```python
import matplotlib.pyplot as plt

plt.xlabel("Age")
plt.ylabel("Number of students")
plt.title("Histogram of age")

plt.hist(df['Age'],edgecolor='black')

plt.show()
```

```python
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title("Scattered plot of Age and Salary")
plt.scatter(df['Age'],df['Salary'],color='cyan')

plt.show()
```

```python
df['Discounted_Price'] = df['Price'] - (df['Price'] * 0.1)
df
```

Scattered Plot - Price Vs Discounted Price - Positive Correlation

```python
plt.xlabel("Discounted Price")
plt.ylabel("Price")
plt.title("Scattered plot of Price Vs Discounted Price")

plt.scatter(df["Price"],df["Discounted_Price"],color='purple')
plt.show()
```
