```python
import pandas as pd

df = pd.read_csv("model_data.csv")
df.sample(5)
```

Positive correlation

```python
import matplotlib.pyplot as plt

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Correlation between Study hours and Exam score")

plt.scatter(df['study_hours'],df['exam_score'],marker='*',color='green')
```

Negative correlation

```python
import matplotlib.pyplot as plt

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Correlation between Social media hours and Exam score(Negative)")

plt.scatter(df['social_media_hours'],df['exam_score'],color='red',marker='x')
```

No correlation

```python
import matplotlib.pyplot as plt

plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.title("Correlation between Favorite number and Exam score(No relation)")

plt.scatter(df['favorite_number'],df['exam_score'],color='#1494b8')
```
