Filter integers divisible by 3 but not divisible by 5

```python
nums = list(map(int,input("Enter numbers: ").split()))

filtered = list(filter(lambda x: x % 3 == 0 and x % 5 != 0, nums))

print(filtered)
```

Write a function compute(a, b) that:
Returns (a² + b²) / (a - b)

```python
def comput(a,b):
    try:
        divider = a-b;
        result = (a**2+b**2)/divider
        return result
    except ZeroDivisionError:
        print("Error: Divider can not be zero")
        
x,y = list(map(int,input("Enter two values: ").split()))
        
print(comput(x,y))
```

Write a Python program that reads a text file and finds the most frequently occurring word.

```python
try:
    with open("./data.txt",'r') as file:
        content = file.read().lower()
        # print(content)
        
        words = content.split()
        
        word_count = {}
        
        for word in words:
            word_count[word] = word_count.get(word,0) + 1
        
        maxOccuringWord = None
        maxCount = 0
        
        for key,value in word_count.items():
            if value > maxCount:
                maxCount = value
                maxOccuringWord = key
        
        print(maxOccuringWord)
except Exception as e:
    print(e)
```

Write a program that converts a sentence into a list representing the length of each word. Ignore extra spaces

```python
sentence = input("Enter the sentence: ")

wordsArr = sentence.split()
print(wordsArr)

lengthArr = list(map(lambda word : len(word),wordsArr))
print(lengthArr)
```

Mirror Number

```python
num = int(input("Enter the number: "))

original_num = num;
reversed_num = 0;

while num != 0:
    singleDigit = num % 10;
    reversed_num = reversed_num * 10 + singleDigit
    num = num // 10

if(reversed_num == original_num):
    print("Mirror Number")
else:
    print("Not Mirror Number")
```
