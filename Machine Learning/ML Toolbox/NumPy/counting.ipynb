```python
import numpy as np

arr = np.random.randint(1,100,(100))
# or arr = np.random.randint(1,100,100)
# np.random.randint(low, high(not included), size)
print(arr)

print("-------------------")
anotherArr = np.random.randint(1,10,7)
print("Another array with 9 items: ",anotherArr)

print("-------------------")
oneMoreArray = np.random.randint(1,10,(3,4));
# an array with randnom numbers ranging from 1 to 9, in a 3*4 matrix form
print("One more array as matrix:\n",oneMoreArray)

print("=======================")

greather_than_50 = np.count_nonzero(arr > 50)
print(greather_than_50)
```

```python
unique_val = np.unique(arr)
print(unique_val)
```

```python
unique_val = np.unique(arr, return_counts=True)
print(unique_val)
```

```python
unique_val, count = np.unique(arr, return_counts=True)
print(unique_val)
print(count)
```
