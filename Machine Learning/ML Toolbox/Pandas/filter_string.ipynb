```python
import pandas as pd

data = {
    "Name": [
        "Jhon Doe", "David Ross", "Sarah Chen", "Mike Ross", "Emily White",
        "Liam O'Connor", "Sofia Rodriguez", "Aisha Khan", "Yuki Tanaka",
        "Kwame Mensah", "Elena Petrova", "Jack Harrison"
    ],
    "City": [
        "North York", "New York", "Toronto", "Toronto", "London",
        "Dublin", "Madrid", "Dubai", "Tokyo",
        "Accra", "Berlin", "Sydney"
    ],
    "Country": [
        "Canada", "US", "Canada", "Canada", "UK",
        "Ireland", "Spain", "UAE", "Japan",
        "Ghana", "Germany", "Australia"
    ],
    "Job Title": [
        "Software Engineer", "Backend Developer", "Data Scientist", "Project Manager", "UX Designer",
        "DevOps Engineer", "Frontend Developer", "Product Manager", "Systems Architect",
        "Data Analyst", "Security Engineer", "QA Engineer"
    ],
    "Salary": [
        160000, 155000, 145000, 130000, 120000,
        140000, 115000, 170000, 180000,
        95000, 135000, 110000
    ]
}

df = pd.DataFrame(data)

df
```

simple string filtering

```python
df.loc[df['City'].str.contains('York',case='False')]
```

### string filtering with regex

all cities starting with T

```python
df.loc[df['City'].str.contains(r'T')]
```

all cities ending with 'n'

```python
df.loc[df['City'].str.contains(r'n$')]
```

All Developers and Engineers

```python
df.loc[df['Job Title'].str.contains(r'Developer|Engineer')] # be careful about white space, it counts white spaces as well
```

All names that start with a vowel

```python
df.loc[df['Name'].str.contains(r'^[AEIOU]')]
```
