```python
import pandas as pd

df = pd.read_csv('student_completed_data.csv')

df.sample(5)
```

```python
instructor_group = df.groupby('Instructor')
instructor_group
```

```python
for instructor, df_instruct in instructor_group:
    print(instructor)
    display(df_instruct)
```

```python
instructor_group.max()
```

```python
instructor_group.min()
```

```python
instructor_group.describe()
```

```python
mean_marks = instructor_group['Total Marks'].mean()
mean_marks
```

```python
# replacing NaN values of Total Marks by group wise average mark
df['Total Marks'] = df['Total Marks'].fillna(df['Instructor'].map(mean_marks))

df
```
