```python
import pandas as pd

df = pd.read_csv("student_data.csv")
df.sample(3)

status_group = df.groupby('CompletionStatus').size()
status_group
```

```python
import matplotlib.pyplot as plt

plt.pie(status_group,labels=status_group.index,autopct='%1.1f%%',explode=(0,0,0.1),colors = ['green','orange','red'])
plt.legend()
plt.show()
```

```python
location_group = df.groupby('Location').size()
plt.pie(location_group,labels=location_group.index,autopct='%1.1f%%')

plt.show()
```
