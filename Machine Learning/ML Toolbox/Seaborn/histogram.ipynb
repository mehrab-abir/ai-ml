```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')
tipsdf.sample(5)
```

Axes level

```python
sns.histplot(data=tipsdf, x='day')
```

```python
sns.histplot(data=tipsdf, x='day', hue='sex')
```

Figure level

```python
sns.displot(kind='hist',data=tipsdf, x='day',col='sex')
```

```python
sns.displot(kind='hist',data=tipsdf, x='sex',col='day',col_wrap=2)
```

```python
sns.displot(kind='hist',data=tipsdf, x='total_bill',col='sex')
```

```python
sns.displot(kind='hist',data=tipsdf, x='tip',col='sex')
```

```python
sns.displot(kind='hist',data=tipsdf, x='tip',col='day',col_wrap=2)
```
