```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')
tipsdf.sample(5)
```

Regplot helps decide whether linear regression algorithm can be applied for prediction

```python
sns.regplot(data=tipsdf,x='total_bill',y='tip')
```

lmplot (Lmplot) .. let us use hue

```python
sns.lmplot(data=tipsdf,x='total_bill',y='tip',hue='sex')
```
