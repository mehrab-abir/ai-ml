```python
import pandas as pd

stdf = pd.read_csv("datafiles/sns_data.csv")
stdf.sample(5)
```

```python
import seaborn as sns

sns.scatterplot(data=stdf,x='study_hours',y='test_score')
```

### **Hue** 
- change color based on gender - (differentiate by gender into the plot)

```python
sns.scatterplot(data=stdf,x='study_hours',y='test_score',hue='gender')
```

### **hue** 
- change color based on class_level -(differentiate using class_level)

```python
sns.scatterplot(data=stdf,x='study_hours',y='test_score',hue='class_level')
```

### **style** 
- changes marker based on a column, here -> class_level

```python
sns.scatterplot(data=stdf,x='study_hours',y='test_score',style='class_level')
```

### **size** 
- changes the size of the points based on a column -(differentiate using 'Tshirt_size')

```python
sns.relplot(kind='scatter',data=stdf,x='study_hours',y='test_score',size='Tshirt_size')
```

change the size of the points based on class_level

```python
sns.relplot(kind='scatter',data=stdf,x='study_hours',y='test_score',size='class_level')
```

change the size of the points based on gender

```python
sns.relplot(kind='scatter',data=stdf, x='study_hours',y='test_score',style='gender')
```

### Using 'tips' datasets

```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')

tipsdf.sample(10)
```

changing the color of the points based on 'sex'

```python
sns.scatterplot(data=tipsdf,x='total_bill',y='tip',hue='sex')
```

changing the style of points based 'sex'

```python
sns.relplot(kind='scatter',data=tipsdf,x='total_bill',y='tip',style='sex')
```

changing the color of the points based meal time

```python
sns.relplot(kind='scatter',data=tipsdf,x='total_bill',y='tip',hue='time')
```

changing the style of the points based on meal time

```python
sns.relplot(kind='scatter',data=tipsdf,x='total_bill',y='tip',style='time')
```

changing the size of the points based on meal time

```python
sns.relplot(kind='scatter',data=tipsdf,x='total_bill',y='tip',size='time')
```

multiple paramters in one plot

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill', y='tip',hue='sex',size='size',style='time')
```
