```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("diabetes.csv")
df.sample(5)
```

```python
df.isnull().sum()
```

```python
df[['Glucose','BloodPressure','BMI','SkinThickness']] = df[['Glucose','BloodPressure','BMI','SkinThickness']].replace(0,np.nan)
```

```python
df.sample(5)
```

```python
sns.countplot(data=df,x='Pregnancies')
```

```python
sns.kdeplot(data=df,x='Glucose')
```

```python
sns.kdeplot(data=df,x='BloodPressure')
```

```python
sns.scatterplot(data=df,x='Age',y='BloodPressure')
```

```python
sns.histplot(data=df,x='BMI')
```

```python
from sklearn.impute import SimpleImputer

bmi_imputer = SimpleImputer(missing_values=np.nan,strategy='median')
bmi_imputer.fit(df[['BMI']])

df['BMI'] = bmi_imputer.transform(df[['BMI']])

#! In real project/model training, train/test split happens first, before imputing, then only fit in traiming data, then transform train and test
```
