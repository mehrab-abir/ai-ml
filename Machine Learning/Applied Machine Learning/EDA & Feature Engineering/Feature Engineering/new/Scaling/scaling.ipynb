```python
import pandas as pd

x_train = pd.read_csv("./x_train.csv",index_col=0)
x_test = pd.read_csv("./x_test.csv",index_col=0)
y_train = pd.read_csv("./y_train.csv",index_col=0)
y_test = pd.read_csv("./y_test.csv",index_col=0)
```

### StandardScaler - - for data with normal distribution

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler();

scaler.fit(x_train[['Age']]);

x_train['Age'] = scaler.transform(x_train[['Age']]);
x_test['Age'] = scaler.transform(x_test[['Age']]);
```

```python
import seaborn as sns

sns.kdeplot(data=x_train, x='Age')
```

### Min Max Scaler <br>
0-1 scaler

```python
from sklearn.preprocessing import MinMaxScaler

minMaxScaler = MinMaxScaler()

minMaxScaler.fit(x_train[['Fare']])

x_train['Fare'] = minMaxScaler.transform(x_train[['Fare']])
x_test['Fare'] = minMaxScaler.transform(x_test[['Fare']])
```
