```python
import pandas as pd

x_train = pd.read_csv("x_train.csv", index_col=0)
x_test = pd.read_csv("x_test.csv", index_col=0)
```

```python
x_train.head(5)
```

### **Ordinal encoding - for data with an order**

```python
from sklearn.preprocessing import OrdinalEncoder

pclass_encoder = OrdinalEncoder(categories=[['third','second','first']])

# fit
pclass_encoder.fit(x_train[['Pclass']]);
x_train['encoded_pclass'] = pclass_encoder.transform(x_train[['Pclass']]).ravel();

# test data
x_test['encoded_pclass'] = pclass_encoder.transform(x_test[['Pclass']]).ravel();
```

```python
x_train.drop('Pclass',axis=1,inplace=True)
```

```python
x_train.sample(5)
# x_test.sample(5)
```

```python
x_test.drop('Pclass',axis=1,inplace=True)
```

```python
x_test.sample(2)
```

### **Nominal encoding - For data with no order** <br>
**One hot encoding**

```python
from sklearn.preprocessing import OneHotEncoder

sex_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas');

sex_ohe.fit(x_train[['Sex']])

# transform
encoded_sex_df_train = sex_ohe.transform(x_train[['Sex']])
encoded_sex_df_test = sex_ohe.transform(x_test[['Sex']])

# print(encoded_sex_df_train)

x_train = pd.concat([x_train,encoded_sex_df_train],axis=1);
x_test = pd.concat([x_test,encoded_sex_df_test],axis=1);
```

```python
x_train.drop(columns='Sex',inplace=True)
x_test.drop(columns='Sex',inplace=True)
```

```python
# x_test.sample(3)
x_train.sample(3)
```

For 'Embarked' column

```python
embarked_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas');

embarked_ohe.fit(x_train[['Embarked']])

# transform
encoded_embarked_df_train = embarked_ohe.transform(x_train[['Embarked']])
encoded_embarked_df_test = embarked_ohe.transform(x_test[['Embarked']])

# print(encoded_embarked_df_train)

x_train = pd.concat([x_train,encoded_embarked_df_train],axis=1);
x_test = pd.concat([x_test,encoded_embarked_df_test],axis=1);
```

```python
x_train.drop(columns='Embarked',inplace=True)
x_test.drop(columns='Embarked',inplace=True)
```

```python
# x_test.sample(3)
x_train.sample(3)
```

Encoding 'Cabin' column

```python
# extracting "Deck" name from cabin name
x_train['Cabin_Deck'] = x_train['Cabin'].astype(str).str[0];
x_test['Cabin_Deck'] = x_test['Cabin'].astype(str).str[0];
```

```python
x_train.sample(3)
```

One hot encoding for 'cabin_deck'

```python
from sklearn.preprocessing import OneHotEncoder

deck_ohe = OneHotEncoder(sparse_output=False, drop='first').set_output(transform='pandas')

# fit
deck_ohe.fit(x_train[['Cabin_Deck']])

# transform
encoded_deck_train = deck_ohe.transform(x_train[['Cabin_Deck']])
encoded_deck_test = deck_ohe.transform(x_test[['Cabin_Deck']])

# concat with df
x_train = pd.concat([x_train,encoded_deck_train],axis=1)
x_test = pd.concat([x_test,encoded_deck_test],axis=1)
```

```python
# x_train.sample(4)
x_train.head(4)
```

```python
x_train.drop(columns='Cabin',inplace=True);
x_test.drop(columns='Cabin',inplace=True);
```

### **Label Encoder - encoding the Target column**

```python
y_train = pd.read_csv("y_train.csv", index_col=0).squeeze("columns")
y_test = pd.read_csv("y_test.csv", index_col=0).squeeze("columns")
y_train.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder();

label_encoder.fit(y_train);

y_train_array = label_encoder.transform(y_train)
y_test_array = label_encoder.transform(y_test)
```

```python
print(y_train_array)
```

```python
# converting the array to pandas series
y_train = pd.Series(y_train_array,index=y_train.index,name=y_train.name)
y_test = pd.Series(y_test_array,index=y_test.index,name=y_test.name)
```
