```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

exp_x = np.array([1,2,4,7,8,10])
sal_y = np.array([45,50,62,82,95,110])
```

```python
plt.scatter(exp_x,sal_y)
plt.xlabel("Exp.")
plt.ylabel("Salary(K)")
```

### **y = wx + b**<br>
w = weights (m in y = mx + c)<br>
b = bias (c in y = mx + c)<br>
x = input(feature)<br>
y = predicted output(target)

```python
# a simple predicting function

def makePrediction(x,y,w,b):
    m = x.shape[0] # total number of rows in a feature
    
    prediction_list = np.zeros(m,)
    for i in range(m):
        prediction_list[i] = w * x[i] + b
    
    return prediction_list;
```

w = 0, b = 0

```python
prediction_list = makePrediction(exp_x,sal_y,0,0) # w = 0, b = 0
print(prediction_list)

plt.scatter(exp_x,sal_y)
plt.plot(exp_x,prediction_list,color="purple")
plt.xlabel("Exp.")
plt.ylabel("Salary(K)")
```
