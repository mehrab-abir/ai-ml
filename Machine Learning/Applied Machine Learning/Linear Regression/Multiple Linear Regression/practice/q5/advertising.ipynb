```python
import pandas as pd
import numpy as np

df = pd.read_csv("advertising_data.csv")
```

```python
df.sample(5)
```

```python
data_array = df[['TV','radio','newspaper','sales']].to_numpy()
print(data_array)
```

```python
x = data_array[:,0:3];
y = data_array[:,3];
```

```python
def make_prediction(x,w,b):
    rows = x.shape[0];
    prediction_list = np.zeros(rows,);
    
    for i in range(rows):
        prediction_list[i] = np.dot(w,x[i])+b;
    
    return prediction_list;
```

```python
rows = x.shape[0];
cols = x.shape[1];

w_init = np.ones(cols,);
b_init = 1;

print(make_prediction(x,w_init,b_init))
```

```python
def compute_cost(x,y,w,b):
    m = x.shape[0];
    
    predictions = make_prediction(x,w,b);
    cost = (np.sum((predictions-y)**2))/(2*m);

    return cost;
```

```python
w_init = np.ones(cols,);
b_init = 1;
print(compute_cost(x,y,w_init,b_init))
```

```python
def calculate_gradient(x,y,w,b):
    nRows = x.shape[0];
    nCols = x.shape[1];
    
    dj_dw = np.zeros(nCols,);
    dj_db = 0.0;
    
    for row in range(nRows):
        prediction = np.dot(w,x[row]) + b;
        error = prediction - y[row];
        dj_db = dj_db + error;
        
        for col in range(nCols):
            dj_dw[col] = dj_dw[col] + (error*x[row,col]);
        
    dj_dw = dj_dw/nRows;
    dj_db = dj_db/nRows;
    
    return dj_dw, dj_db;
        
```

```python
nRows = x.shape[0];
nCols = x.shape[1];

w_init = np.zeros(nCols,);
b_init = 1.0;

print(calculate_gradient(x,y,w_init,b_init))
```

```python
def gradient_descent(x,y,w_init,b_init,alpha,max_iter):
    w = w_init;
    b = b_init;
    
    cost_memo = [];
    iteration = [];
    
    for i in range(max_iter):
        dj_dw, dj_db = calculate_gradient(x,y,w,b);
        
        w = w - (alpha*dj_dw);
        b = b - (alpha*dj_db);
        
        cost = compute_cost(x,y,w,b);
        cost_memo.append(cost);
        iteration.append(i);
        
        if(i%100 == 0):
            print(f"i={i}, cost: {cost:0.4f}")
    
    return w,b,cost_memo,iteration;
    
```

```python
nRows = x.shape[0]; #? number of rows
nCols = x.shape[1]; #? number of columns

w_init = np.zeros(nCols,);
b_init = 0.0;

w_final, b_final, cost_memo, iter = gradient_descent(x,y,w_init,b_init,alpha=0.0000011, max_iter=1000)
```
