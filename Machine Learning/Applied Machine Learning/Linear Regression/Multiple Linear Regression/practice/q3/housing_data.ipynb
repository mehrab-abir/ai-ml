```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing()
```

```python
x = housing.data[:,0:3]
features = housing.feature_names # features names
y = housing.target

print("Features:\n",features[0:3])
print(x)
print("Target:\n",housing.target_names)
print(y)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x,y);

weights = model.coef_;
bias = model.intercept_;

print(f"Weights: {weights}\nBias: {bias}")
```

```python
import numpy as np

def make_prediction(x,w,b):
    m = x.shape[0];
    prediction_list = np.zeros(m,);
    
    for i in range(m):
        prediction_list[i] = np.dot(w,x[i])+b;
    
    return prediction_list;
```

```python
print(f"Actual values:\n{y[0:5]}")
predicted_values = make_prediction(x,weights,bias)
print(f"Predicted values:\n{predicted_values[0:5]}")
```
