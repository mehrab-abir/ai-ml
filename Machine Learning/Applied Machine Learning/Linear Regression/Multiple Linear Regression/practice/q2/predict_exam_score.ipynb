```python
import numpy as np

x = np.array([
    [6, 10, 60, 70],
    [2, 7, 50, 45],
    [8, 8, 90, 88],
    [4, 6, 70, 60],
    [9, 9, 95, 92],
    [1, 5, 40, 35],
    [5, 7, 80, 75],
    [7, 8, 85, 82],
    [3, 6, 60, 50],
    [10, 8, 100, 96]
])
y = x[:,3];
```

```python
def make_prediction(x,w,b):
    m = x.shape[0];
    prediction_list = np.zeros(m,)
    
    for i in range(m):
        prediction_list[i] = np.dot(w,x[i]) + b;
        
    return prediction_list;
```

```python
m = x.shape[0];
n = x.shape[1];

w = np.ones(n,);
b = 2;

print("Acutal values: ",x[:,3]);
print("Predicted values: ",make_prediction(x,w,b))
```

```python
def compute_cost(x,y,w,b):
    m = x.shape[0];
    
    predictions = make_prediction(x,w,b);
    cost = (np.sum((predictions-y)**2))/(2*m);
    
    return cost;
```

```python
w_init = np.ones(n,);
b_init = 2;

print(compute_cost(x,y,w_init,b_init))
```

```python
w_init[0] = 10;
print(compute_cost(x,y,w_init,b_init))
```

```python
cost_tracked_w = [];
for i in range(-10,11,1):
    w_init[0] = i;
    cost_tracked_w.append(compute_cost(x,y,w_init,b_init))

print(cost_tracked_w)
```

```python
import matplotlib.pyplot as plt

w_0 = np.arange(-10,11,1)

plt.plot(w_0,cost_tracked_w)
plt.xlabel("w[0] from -10 to 10")
plt.ylabel("change of cost")
```
