Simple Linear Regression uses one feature (independent variable) to predict one target variable (dependent variable).<br>
In ML, regression means: Predicting a continuous numeric value.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

exp_x = np.array([1,2,4,7,8,10])
sal_y = np.array([45,50,62,82,95,110])
```

```python
plt.scatter(exp_x,sal_y)
plt.xlabel("Exp.")
plt.ylabel("Salary(K)")
```

### **y = wx + b**<br>
w = weights (m in y = mx + c)<br>
b = bias (c in y = mx + c)<br>
x = input(feature)<br>
y = predicted output(target)

```python
# a simple predicting function

def makePrediction(x,y,w,b):
    m = x.shape[0] # total number of rows in a feature
    
    prediction_list = np.zeros(m,)
    for i in range(m):
        prediction_list[i] = w * x[i] + b
    
    return prediction_list;
```

### Cost Function
J(m,b)=∑i=1n​(yi​−y^​i​)2 <br> <br>
(y1 - yp1)^2 + (y2 - yp2)^2 + (y3 - yp3)^2 + ... + (yn - ypn`)^2<br>
must be minimized
- Mean Squared Error

```python
def compute_cost(x,y,w,b):
    m = x.shape[0]
    
    cost = 0.0 # initializing cost with 0.0
    
    for i in range(m):
        prediction = w * x[i] + b;
        error = prediction - y[i];
        error_squared = error ** 2;
        cost = cost + error_squared;
    
    cost = cost/(2*m); # m = number of rows, --> mean squared error
    #? divide by '2m' instead of 'm' for convenience in gradient descent calculation
    
    return cost;
```

```python
w = 0.0;
b = 0.0;

prediction_list = makePrediction(exp_x,sal_y,w=w,b=b) #* w = 0, b = 0
print(prediction_list)

cost = compute_cost(exp_x,sal_y,w=w,b=b)

plt.scatter(exp_x,sal_y)
plt.plot(exp_x,prediction_list,color="purple")
plt.title(f"Cost = {cost}, when w={w}, b={b}")
plt.xlabel("Exp.")
plt.ylabel("Salary(K)")
plt.show()
```

### **Gradient Descent Algorithm**

dj_dw = 1/m * (summation of (yp-y) * x[i])...n<br>
dj_db = 1/m * (summation of (yp-y))...n<br>
<br>
- yp-y = error, yp = prediction, y = actual value

w = w(initial) - (alpha * dj_dw)...[alpha = 0.01]<br>
b = b(initial) - (alpha * dj_db)<br>

```python
def calculate_gradient(x,y,w,b):
    m = x.shape[0];
    
    dj_dw = 0.0; # derivative of the cost function with respect to w (weights)
    dj_db = 0.0; # derivative of the cost function with respect to b (bias)
    
    for i in range(m):
        prediction = w * x[i] + b;
        error = prediction - y[i];
        dj_dw = dj_dw + (error * x[i]);
        dj_db = dj_db + error
    
    dj_dw = dj_dw/m;
    dj_db = dj_db/m;
    
    return dj_dw, dj_db;
```

```python
print(calculate_gradient(exp_x,sal_y,w=8, b=4)); # dj_dw, dj_db
```

```python
def gradient_descent(x,y,w_init,b_init,max_iter, alpha=0.01):
    w = w_init;
    b = b_init;
    
    cost_memo = [];
    iteration = [];
    
    for i in range(max_iter):
        dj_dw, dj_db = calculate_gradient(x,y,w,b);
        
        # updating w and b
        w = w - (alpha * dj_dw);
        b = b - (alpha * dj_db);
        
        cost = compute_cost(x,y,w,b)
        
        cost_memo.append(cost);
        iteration.append(i);
        
        if(i%100 == 0):
            print(f"iteration: {i} -: w = {w:0.4f}, b = {b:0.4f}, dj_dw = {dj_dw:0.4f}, dj_db = {dj_db:0.4f}, cost = {cost:0.4f}")
    return w,b,cost_memo, iteration
```

```python
w_final,b_final,cost_memo, iter = gradient_descent(exp_x,sal_y,w_init=0.0,b_init=0.0,max_iter=10000,alpha=0.01)
```

Ploting prediction line with final values of 'w' and 'b' and minimized cost

```python
w = w_final;
b = b_final;

prediction_list = makePrediction(exp_x,sal_y,w=w,b=b) #* w = 0, b = 0
# print(prediction_list)

cost = compute_cost(exp_x,sal_y,w=w,b=b)

plt.scatter(exp_x,sal_y)
plt.plot(exp_x,prediction_list,color="green")
plt.title(f"Cost = {cost:0.3f}, when w={w:0.3f}, b={b:0.3f}")
plt.xlabel("Exp.")
plt.ylabel("Salary(K)")
plt.show()
```
