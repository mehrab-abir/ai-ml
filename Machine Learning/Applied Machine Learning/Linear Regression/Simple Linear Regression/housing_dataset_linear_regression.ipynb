```python
import pandas as pd

df = pd.read_csv("Housing.csv")
df.head()
```

```python
import numpy as np

area = np.array(df['area'])
price = np.array(df['price'])
print("Area: ",area[:10])
print("Price: ", price[:10])
```

```python
import matplotlib.pyplot as plt

plt.scatter(area,price)
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")
plt.show()
```

```python
def make_prediction(x,y,w,b):
    m = x.shape[0]
    
    prediction_list = np.zeros(m,);
    for i in range(m):
        prediction_list[i] = w*x[i] + b;
        
    return prediction_list;
```

```python
prediction_list = make_prediction(area,price,w=1000,b=2000);

plt.scatter(area,price)
plt.plot(area,prediction_list,color='green')
plt.xlabel("Area")
plt.ylabel("Price")
plt.title("Area vs Price")
plt.show()
```

Cost - Mean Squared Error

```python
def compute_cost(x,y,w,b):
    m = x.shape[0];
    
    cost = 0.0;
    
    for i in range(m):
        prediction = w * x[i] + b;
        error = prediction - y[i];
        error_squared = error ** 2;
        cost = cost + error_squared;
    
    cost = cost/(2*m);
    
    return cost;
```

```python
print(f"w = 1000, b = 2000: {compute_cost(area,price,w=1000,b=2000)}")
print(f"w = 600, b = 1500: {compute_cost(area,price,w=600,b=1500)}")
print(f"w = 300, b = 1000: {compute_cost(area,price,w=300,b=1000)}")
print(f"w = 70, b = 50: {compute_cost(area,price,w=20,b=10)}")
```

```python
w_range = [];
cost_history = [];

for w_i in range(-1000,3000):
    cost_i = compute_cost(area,price,w_i,0.0);
    cost_history.append(cost_i)
    w_range.append(w_i);

plt.plot(w_range,cost_history);
plt.xlabel("Weights")
plt.ylabel("Cost")
plt.title("weights vs cost - (b -> constant)")
plt.show()
```

Gradient Descent

```python
def calculate_gradient(x,y,w,b):
    m = x.shape[0];
    
    dj_dw = 0.0;
    dj_db = 0.0;
    
    for i in range(m):
        prediction = w * x[i] + b;
        error = prediction -  y[i];
        dj_dw = dj_dw + (error * x[i]);
        dj_db = dj_db + error;
    
    dj_dw = dj_dw/m;
    dj_db = dj_db/m;
    
    return dj_dw, dj_db;
```

```python
print(calculate_gradient(area,price,w=500,b=600))
```

```python
def gradient_descent(x,y,w_init,b_init,max_iter,alpha):
    w = w_init;
    b = b_init;
    
    cost_memo = [];
    iteration = [];
    
    for i in range(max_iter):
        dj_dw, dj_db = calculate_gradient(x,y,w,b);
        
        w = w - (alpha*dj_dw);
        b = b - (alpha*dj_db);
        
        cost = compute_cost(x,y,w,b);
        
        cost_memo.append(cost);
        iteration.append(i);
        
        if(i%100 == 0):
            print(f"iteration {i}: w={w:0.3f},b={b:0.3f}, cost={cost:0.3f}")
    
    return w,b,cost_memo,iteration;
```

```python
w_final, b_final, cost_memo, iteration = gradient_descent(area,price,0.0,0.0,10000,0.000000000001)
print(f"W_Final = {w_final}")
print(f"B_Final = {b_final}")
```

```python
plt.plot(iteration,cost_memo)
plt.xlabel("Number of iteration")
plt.ylabel("Cost")
```
