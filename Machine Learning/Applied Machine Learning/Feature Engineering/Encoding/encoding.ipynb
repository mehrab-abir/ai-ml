Encoding:
- Ordinary - There is an order in data, {First class, Second class, Third Class}, {MSc,BSc, HSC} etc.
- Nominal - no order in data, Gender, Color, Type etc.

```python
import pandas as pd

x_train = pd.read_csv("titanic_x_train.csv")
x_test = pd.read_csv("titanic_x_test.csv")
x_train.drop(['z_score_age','z_score_fare'],axis=1,inplace=True)
```

```python
x_train.sample(4)
```

```python
x_test.sample(3)
```

Ordinal Encoder - encode categorical features as an integer array

```python
from sklearn.preprocessing import OrdinalEncoder
pclass_encoder = OrdinalEncoder(categories=[['third','second','first']]) # in ascending order 

# fit --for training data only
pclass_encoder.fit(x_train[['Pclass']])

# transform -- for both training data and test data
x_train['Pclass_encoded'] = pclass_encoder.transform(x_train[['Pclass']]).ravel()
x_test['Pclass_encoded'] = pclass_encoder.transform(x_test[['Pclass']]).ravel()
```

Deleting 'Pclass' column as it has been encoded

```python
x_train.drop('Pclass',axis=1,inplace=True)
```

```python
x_train.sample(5)
```

```python
x_test.drop('Pclass',axis=1,inplace=True)
```

```python
x_test.sample(3)
```

One hot encoder - encoding Nominal Data. <br>
Encode categorical features as a one-hot numeric array.

OneHotEncoding for 'Sex' column

```python
from sklearn.preprocessing import OneHotEncoder

sex_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

# fit
sex_ohe.fit(x_train[['Sex']])

# transform
encoded_sex_df_train = sex_ohe.transform(x_train[['Sex']])
encoded_sex_df_test = sex_ohe.transform(x_test[['Sex']])

# concatenating encoded_sex_df with x_train and x_test df
x_train = pd.concat([x_train,encoded_sex_df_train],axis=1)
x_test = pd.concat([x_test,encoded_sex_df_test],axis=1)
```

```python
x_train.sample(2)
```

```python
# dropping 'Sex' column as it is now encoded
x_train.drop(columns='Sex',inplace=True)
x_test.drop('Sex',axis=1,inplace=True)

# both ways of deleting columns are correct
```

OneHotEncoding for 'Embarked' column

```python
embarked_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

# fit
embarked_ohe.fit(x_train[['Embarked']])

# transform
encoded_embarked_df_train = embarked_ohe.transform(x_train[['Embarked']])
encoded_embarked_df_test = embarked_ohe.transform(x_test[['Embarked']])

# concatenating df's
x_train = pd.concat([x_train,encoded_embarked_df_train],axis=1)
x_test = pd.concat([x_test,encoded_embarked_df_test],axis=1)
```

```python
# dropping "Embarked" column as it is now encoded
x_train.drop(columns='Embarked',inplace=True)
x_test.drop(columns='Embarked',inplace=True)
```
