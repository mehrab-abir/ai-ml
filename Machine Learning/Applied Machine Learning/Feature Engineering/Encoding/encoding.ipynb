Encoding:
- Ordinary - There is an order in data, {First class, Second class, Third Class}, {MSc,BSc, HSC} etc.
- Nominal - no order in data, Gender, Color, Type etc.

```python
import pandas as pd

# index_col = 0 --> restored original index
x_train = pd.read_csv("titanic_x_train.csv",index_col = 0)
x_test = pd.read_csv("titanic_x_test.csv",index_col = 0)
x_train.drop(['z_score_age','z_score_fare'],axis=1,inplace=True)
```

```python
x_train.head(4)
```

```python
x_test.sample(3)
```

Ordinal Encoder - encode categorical features as an integer array

```python
from sklearn.preprocessing import OrdinalEncoder
pclass_encoder = OrdinalEncoder(categories=[['third','second','first']]) # in ascending order 

# fit --for training data only
pclass_encoder.fit(x_train[['Pclass']])

# transform -- for both training data and test data
x_train['Pclass_encoded'] = pclass_encoder.transform(x_train[['Pclass']]).ravel()
x_test['Pclass_encoded'] = pclass_encoder.transform(x_test[['Pclass']]).ravel()
```

Deleting 'Pclass' column as it has been encoded

```python
x_train.drop('Pclass',axis=1,inplace=True)
```

```python
x_train.head(5)
```

```python
x_test.drop('Pclass',axis=1,inplace=True)
```

```python
x_test.head(3)
```

One hot encoder - encoding Nominal Data. <br>
Encode categorical features as a one-hot numeric array.

OneHotEncoding for 'Sex' column

```python
from sklearn.preprocessing import OneHotEncoder

sex_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

# fit
sex_ohe.fit(x_train[['Sex']])

# transform
encoded_sex_df_train = sex_ohe.transform(x_train[['Sex']])
encoded_sex_df_test = sex_ohe.transform(x_test[['Sex']])

# concatenating encoded_sex_df with x_train and x_test df
x_train = pd.concat([x_train,encoded_sex_df_train],axis=1)
x_test = pd.concat([x_test,encoded_sex_df_test],axis=1)
```

```python
x_train.sample(2)
```

```python
# dropping 'Sex' column as it is now encoded
x_train.drop(columns='Sex',inplace=True)
x_test.drop('Sex',axis=1,inplace=True)

# both ways of deleting columns are correct
```

```python
# x_test.sample(2)
```

OneHotEncoding for 'Embarked' column

```python
embarked_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

# fit
embarked_ohe.fit(x_train[['Embarked']])

# transform
encoded_embarked_df_train = embarked_ohe.transform(x_train[['Embarked']])
encoded_embarked_df_test = embarked_ohe.transform(x_test[['Embarked']])

# concatenating df's
x_train = pd.concat([x_train,encoded_embarked_df_train],axis=1)
x_test = pd.concat([x_test,encoded_embarked_df_test],axis=1)
```

```python
# dropping "Embarked" column as it is now encoded
x_train.drop(columns='Embarked',inplace=True)
x_test.drop(columns='Embarked',inplace=True)
```

```python
# x_train.sample(3)
# x_test.sample(3)
```

### Encoding the 'Cabin' column

```python
# making a column with the first Character of Cabin name which is 'deck' 
x_train['Cabin_deck'] = x_train['Cabin'].astype(str).str[0];
x_test['Cabin_deck'] = x_test['Cabin'].astype(str).str[0];
```

```python
# x_train.sample(5)
# x_test.sample(5)
```

Now, one hot encoding for 'Cabin_deck' column

```python
deck_ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')

# fit
deck_ohe.fit(x_train[['Cabin_deck']])

# transform
encoded_deck_df_train = deck_ohe.transform(x_train[['Cabin_deck']])
encoded_deck_df_test = deck_ohe.transform(x_test[['Cabin_deck']])

# concate df
x_train = pd.concat([x_train, encoded_deck_df_train],axis=1)
x_test = pd.concat([x_test, encoded_deck_df_test],axis=1)
```

```python
x_test.head(10)
```

```python
x_train.drop(columns='Cabin_deck',axis=1,inplace=True)
x_test.drop(columns='Cabin_deck',axis=1,inplace=True)
```

### **Label Encoder** - encoding 'Target' column (y_train, y_test)

```python
y_train = pd.read_csv("titanic_y_train.csv",index_col=0).squeeze('columns')
y_test = pd.read_csv("titanic_y_test.csv",index_col=0).squeeze('columns')
y_train.head(5)
```

```python
from sklearn.preprocessing import LabelEncoder

lable_encoder = LabelEncoder()

# fit
lable_encoder.fit(y_train) # no need to pass column name, as it's just a series already(one column only)

# transform
y_train_array = lable_encoder.transform(y_train)
y_test_array = lable_encoder.transform(y_test)
```

Converting y_train and y_test array to pandas series

```python
y_train = pd.Series(y_train_array,index=y_train.index,name=y_train.name)
y_test = pd.Series(y_test_array,index=y_test.index,name=y_test.name)
```
