Encoding:
- Ordinary - There is an order in data, {First class, Second class, Third Class}, {MSc,BSc, HSC} etc.
- Nominal - no order in data, Gender, Color, Type etc.

```python
import pandas as pd

x_train = pd.read_csv("titanic_x_train.csv")
x_test = pd.read_csv("titanic_x_test.csv")
x_train.drop(['z_score_age','z_score_fare'],axis=1,inplace=True)
```

```python
x_train.sample(4)
```

```python
x_test.sample(3)
```

Ordinal Encoder - encode categorical features as an integer array

```python
from sklearn.preprocessing import OrdinalEncoder
pclass_encoder = OrdinalEncoder(categories=[['third','second','first']]) # in ascending order 

# fit --for training data only
pclass_encoder.fit(x_train[['Pclass']])

# transform -- for both training data and test data
x_train['Pclass_encoded'] = pclass_encoder.transform(x_train[['Pclass']]).ravel()
x_test['Pclass_encoded'] = pclass_encoder.transform(x_test[['Pclass']]).ravel()
```
