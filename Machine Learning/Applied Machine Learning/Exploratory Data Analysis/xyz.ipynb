```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "City": ["Dhaka", "Dhaka", "CTG", "CTG", "Sylhet"],
    "Salary": [50000, 60000, 45000, 47000, 60000]
})

sns.barplot(data=df, x='City', y='Salary')

plt.title("Average Salary by City")
plt.show()
```
