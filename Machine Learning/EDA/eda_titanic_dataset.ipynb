### 5. Univariate Analysis( Categorical Value Analysis -1 )

```python
df['Survived'].value_counts()
```

```python
# Pracentage
(df['Survived'].value_counts()/len(df))*100
```

```python
survibed_label = df['Survived'].unique()
print(survibed_label)

survibed_count = df['Survived'].value_counts()
print(survibed_count)

plt.pie(survibed_count,labels=survibed_label,autopct="%1.1f%%",colors=['gray','green'])
plt.title("Pie Chart of Survival rate")
plt.show()
```
