```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("penguins")
df.sample(5)
```

```python
sns.countplot(data=df,x=df['species'])
```

```python
sns.countplot(data=df,x=df['island'])
plt.title("Island count")
```

```python
print(df['island'].value_counts())
(df['island'].value_counts()/len(df))*100
```

```python
island_count = (df['island'].value_counts()/len(df))*100
island_label = df['island'].unique()
island_label

plt.pie(island_count,labels=island_label,autopct="%1.1f%%")
plt.show()
```

```python
genderCounts = df['sex'].value_counts()
genderCounts.values
genderCounts.index

plt.pie(genderCounts.values, labels=genderCounts.index,autopct="%1.1f%%")
plt.title("Gender Parcentage")
plt.show()
```
