```python
import pandas as pd

stdf = pd.read_csv("datafiles/sns_data.csv")
stdf.sample(5)
```

```python
import seaborn as sns

sns.lineplot(data=stdf,x='week', y='attendance_rate',errorbar=None)
```

### **Hue**  
- changes color based on a column - (differentiate using 'gender')

```python
sns.lineplot(data=stdf,x='week', y='attendance_rate',errorbar=None,hue='gender')
```

Hue - with class_level, meaning-> change color based on class_level

```python
sns.lineplot(data=stdf,x='week', y='attendance_rate',errorbar=None,hue='class_level')
```

Figure label

```python
sns.relplot(kind='line',data=stdf,x='week', y='attendance_rate',errorbar=None,hue='gender')
```
