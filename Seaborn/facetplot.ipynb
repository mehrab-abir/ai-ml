```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')
tipsdf.sample(5)
```

Facetplot works on Figure level,, needs relplot -- as there are usually multiple plots

col -> divide the data plot based on something,,<br>here, 'sex'

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill', y='tip', col='sex')
```

dividing the data plot based on meal time and sex<br>
'sex' in row-wise, 'meal time' in column wise

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill', y='tip', col='time',row='sex')
```

divide based on 'day',, col-wrap decides the number of plot in one row

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill', y='tip', col='day')
# see the next one - col-wrap
```

dividing based on 'day', 2 plots per row

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill', y='tip', col='day',col_wrap=2)
```

differentiate based on 'sex', show by 'day'

```python
sns.relplot(kind='scatter',data=tipsdf, x='total_bill',hue='sex', y='tip', col='day',col_wrap=2)
```
