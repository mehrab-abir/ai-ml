KDE - Kernel Density Estimation<br>
kdeplot() in Seaborn is used to visualize the distribution (density) of data smoothly.<br>
A smoothed version of a histogram.

```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')
tipsdf.sample(5)
```

```python
sns.kdeplot(data=tipsdf,x='total_bill')

# around $15-$25 has the highest density
```

```python
sns.kdeplot(data=tipsdf,x='tip',hue='sex')
```

```python
sns.kdeplot(data=tipsdf,x='tip',hue='sex',fill=True)
```
