Create a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.

```python
import numpy as np

arr1 = np.array([1,32,3,64,7])
arr2 = np.array([
    [5,2,3,9],
    [6,1,3,4],
    [3,6,7,4]
])

arr3 = np.array([
    [
        [5,2,3,9],
        [6,4,3,4],
        [3,6,1,4]
    ],
    [
        [5,6,3,7],
        [1,4,3,4],
        [3,3,1,4]
    ],
    [
        [1,6,7,8],
        [1,6,2,4],
        [5,3,4,4]
    ]
])

print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)
```

```python
arr_zeros = np.zeros((3,3))
print(arr_zeros)
print("++++++++++++++++")

arr_ones = np.ones((2,5))
print(arr_ones)
print("++++++++++++++++")

arr_seven = np.full((3,3),7)
print(arr_seven)
print("++++++++++++++++")

arr_inf = np.full_like(arr_zeros,np.inf)
print(arr_inf)
```

Create a 4×4 matrix from 0–15. Then , extract:
- the first row</br>
- the last column</br>
- a 2×2 subarray from the center</br>


```python
matrix = np.arange(0,16,1).reshape(4,4)
print(matrix)
print("++++++++++++++++")

firstRow = matrix[0,:]
print(firstRow)
print("++++++++++++++++")

last_col = matrix[:,3:]
print(last_col)
print("++++++++++++++++")
last_col2 = matrix[:,3]
print(last_col2)

print("++++++++++++++++")

sub_arr = matrix[1:3,1:3]
print(sub_arr)
```

Create a 2D array with random integers, replace all values of 3rd column with 1

```python
import numpy as np

mat = np.random.randint(1,20,(4,5))
print(mat)
print('-------------------')

# 3rd column
print(mat[:,2])
mat[:,2] = 0
print('_____________________')
print("3rd column replaced with 0")
print(mat)
print('------------------')
print(mat[:,2])
```

```python
matD = np.random.randint(1,20,(3,3))
print(matD)
print("Diagonal values: ",np.diag(matD))
print('------------------')

np.fill_diagonal(matD,0)
print(matD)
```

### Slicing

a matrix of shape (6,6), get sub-matrix --> 3rd to 5th rows, 2nd to 4th cols

```python
mat = np.random.randint(1,36,(6,6))
print(mat)
print('------------------')
print(mat[2:5,1:3])
```

Row wise sum, column wise sum

```python
mat = np.random.randint(1,6,(4,4))
print(mat)

print('Row wise sum: ')
print(np.sum(mat, axis=1))
print('======================')
print('Column wise sum: ')
print(np.sum(mat, axis=0))
```

Reshape a matrix, if reshaping not possible, print "reshaping not possible"

```python
mat = np.random.randint(1,6,(3,4))
print(mat)
print('======================')

cur_shape = mat.shape
newShape = (4,3)

if(mat.size == newShape[0] * newShape[1]):
    new_mat = mat.reshape(newShape[0],newShape[1])
    print("After reshaping --> ",newShape)
    print(new_mat)
else:
    print("Reshaping not possible")
```
