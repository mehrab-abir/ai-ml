Basic Plot

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

temperature = [4,6,10,12,16,15,14,10,7,5]
time = [7,8,11,12,14,15,17,19,21,22]

#plt.plot(x_axis_data, y_axis_data)
plt.plot(time,temperature)
```

With label

```python
temperature = [4,6,10,12,16,15,14,10,7,5]
time = [7,8,11,12,14,15,17,19,21,22]

plt.xlabel("Daytime")
plt.ylabel("Temperature")
plt.title("Temperature during daytime in Spring")

plt.plot(time,temperature)

plt.show()
```

```python
spring_temperature = [4,6,10,12,13,16,15,14,10,8,7,5]
summer_temperature = [12,14,16,19,23,25,28,26,23,21,17,16]
time = [7,8,11,12,13,14,15,17,19,21,22,23]

plt.xlabel("Daytime")
plt.ylabel("Temperature")
plt.title("Temperature during daytime in Spring and Summer")

plt.plot(time,spring_temperature)
plt.plot(time, summer_temperature)

plt.show()
```
