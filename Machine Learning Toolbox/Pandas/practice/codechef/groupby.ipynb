```python
import pandas as pd
import numpy as np

# Create a sample DataFrame
df = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Value1': [10, 20, 30, 40, 50, 60],
    'Value2': [100, 200, 300, 400, 500, 600]
})
print(df)
```

```python
grouped = df.groupby('Category')
grouped
```

```python
group_min =  grouped.min()
print(group_min)
```

```python
print(grouped['Value1'].min())
```

```python
print(grouped['Value1'].mean())
```

```python
group_max =  grouped.max()
print(group_max)
```

```python
print(grouped['Value2'].max())
```

```python
group_sum =  grouped.sum()
print(group_sum)
```
