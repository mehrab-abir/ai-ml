```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)
```

```python
cleanedA = df.dropna(subset=['A'])
print(cleanedA)
```

Drop rows with less than 2 non-null values

```python
clean_df = df.dropna(thresh=2)
clean_df
```

Drop rows where both Column A and Column B has missing values

```python
clean_AB = df.dropna(subset=['A','B'],how='all')
print(clean_AB)
```
