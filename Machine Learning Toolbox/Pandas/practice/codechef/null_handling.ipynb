```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)  
```

```python
df_filled_A = df.fillna({'A':0,'B':99,'C':"Unknown"})
print(df_filled_A)
```

```python
df_filled_B = df.copy()
df_filled_B['B'] = df_filled_B['B'].ffill()
print(df_filled_B)
```

```python
for column in df.columns:
    if df[column].dtype == 'str':
        df[column] = df[column].fillna("Unknown",inplace=True)
    else:
        df[column] = df[column].fillna(df[column].mean(),inplace=True)

print(df)
```

```python
for column in df.columns:
    if pd.api.types.is_numeric_dtype(df[column]):
        df[column] = df[column].fillna(df[column].mean(),inplace=True)
    else:
        df[column] = df[column].fillna("Unknown",inplace=True)

print(df)
```
