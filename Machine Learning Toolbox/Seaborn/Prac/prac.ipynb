Refer to the 'datafiles' directory for datasets

```python
import pandas as pd
import seaborn as sns

salesdf = pd.read_csv("../datafiles/sales.csv")

salesdf.sample(5)
```

1. Using sales, draw a lineplot of date vs sales

```python
sns.lineplot(data=salesdf,x='date',y='sales')
```

2. From students, plot hours_studied vs marks and color by gender.

```python
stdf = pd.read_csv("../datafiles/student_dataset_complete.csv")
stdf.sample(5)
```

```python
sns.lineplot(data=stdf,x='study_hours',y='Marks',hue='gender')
```

3. Create a grid of scatterplots (hours_studied vs marks) faceted by subject.

```python
sns.relplot(kind='scatter',data=stdf,x='study_hours',y='Marks',col='subject')
```

4. Plot a histogram of marks with a KDE overlay.

```python
sns.kdeplot(data=stdf,x='Marks',fill=True,hue='gender')
```

5. Show the KDE curve for hours_studied.

```python
sns.kdeplot(data=stdf,x='study_hours')
```

6. Display the count of records per subject.

```python
sns.countplot(data=stdf,x='subject',hue='gender')
```

7. Draw a regression line for hours_studied and marks.

```python
sns.regplot(data=stdf,x='study_hours',y='Marks')
```

8. Create a pairplot for students' dataset. 

```python
# usually it is done with a couple of columns, not all columns always
# automatically creates plots for almost all pairs of numeric columns by default.
sns.pairplot(data=stdf)
```
