```python
import seaborn as sns

tipsdf = sns.load_dataset('tips')
tipsdf.sample(5)
```

```python
sns.countplot(data=tipsdf,x='day')
```

```python
sns.countplot(data=tipsdf,x='sex')
```

ploting based on 'day', differentiating data of each day based on 'sex'

```python
sns.countplot(data=tipsdf,x='day',hue='sex')
```

opposite of the previous one

ploting based on 'sex', differentiating each sex based on 'day'

```python
sns.countplot(data=tipsdf,x='sex',hue='day')
```
