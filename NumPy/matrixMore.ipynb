```python
import numpy as np

diagonal_matrix = np.diag([6,3,1])
diagonal_matrix_2 = np.diag([6,3,1,4])

print(diagonal_matrix)
print(diagonal_matrix_2)
```

```python
identity_matrix = np.eye(3,3)
print(identity_matrix)
print("-----------------")
print(np.eye(4,4))
print("___________________________")

# look at the diagon of thses 3 matrix
print(np.eye(3,4))
print("-----------------")
print(np.eye(3,4,1))
print("-----------------")
print(np.eye(3,4,-1))
```

```python
mat_A = np.array([
    [6,4,3],
    [1,2,7],
    [0,5,9]
])

print(mat_A)
print("--------------------")

print(mat_A.T) # mostly used in NumPy
print("--------------------")
print(np.linalg.matrix_transpose(mat_A))
```

```python
print(np.linalg.inv(mat_A))
```
