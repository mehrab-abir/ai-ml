```python
import numpy as np

a = np.array([2,6,4,3,8,5])

# python sort
a.sort() # in-place sorting, modifies the original array

print(a) 

# how to preserve the original array
b = np.array([5,6,3,4,1,9])
x = b.copy()

x.sort()
print(f"x = {x}") # sorted

print(f"b = {b}") # unchanged, as we sorted the copied version
```

### NumPy sorting

```python
a = np.array([3,6,4,1,8,5])

sorted_a = np.sort(a)
print(f"a = {a}")
print(f"sorted array = {sorted_a}")
```

### 2D array sorting

*row wise sorting*

```python
matrix = np.random.randint(1,10,(3,4))
print(matrix)

print("=======================")

row_wise_sorted = np.sort(matrix,axis = 1)

print(row_wise_sorted)
```

*column wise sorting*

```python
matrix = np.random.randint(1,10,(3,4))
print(matrix)
print("=======================")

col_wise_sorted = np.sort(matrix,axis = 0)

print(col_wise_sorted)
```
