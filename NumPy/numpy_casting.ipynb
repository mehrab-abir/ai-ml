```python
import numpy as np

nums = np.array([6,1,5,3.2,4])
print(f"Data type of each item: {nums.dtype}") # casted to float64

nums_int = nums.astype(dtype=np.int32)
print(f"Data type of each item after casting to int32: {nums_int.dtype}") # int32
```

```python
import numpy as np

nums = np.array([6,1,5,3.2,4], dtype=np.int64) # setting dtype during array creation
print(f"Data type of each item: {nums.dtype}") # int64
```
