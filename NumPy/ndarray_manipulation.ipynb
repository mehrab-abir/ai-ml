```python
import numpy as np

arr = np.random.randint(1,100, size=(4,5)) # 10 rows, 5 cols

print(arr)

print(f"Shape : {arr.shape}")

b = arr.reshape(2,10) # must match total number of items (here, 20 and 20)
print(f"Shape : {b.shape}")
```

```python
flattenArr = arr.flatten() # row-wise combined
print(flattenArr)
```

ravel()

```python
print(arr)
print("-------------------------")
colWiseFlatten = np.ravel(arr, order='F') # col wise flatten
rowWiseFlatten = np.ravel(arr, order='C') # row wise flatten

print(colWiseFlatten)
print(rowWiseFlatten)
```

Array concatenate

```python
a = np.random.randint(1,10, size=(3,4))
b = np.random.randint(20,30, size=(3,4))

print(a)
print(b)
print("------------------------------")

# row wise concatenate
row_merged = np.concatenate((a,b),axis=0)

print(row_merged)

print("------------------------------")

# col wise concatenate
col_merged = np.concatenate((a,b),axis=1)

print(col_merged)
```

*row wise concatenate korar somoy column same thakte hobe* </br>
*column wise concatenate korar somoy row same thakte hobe*

**Transpose matrix**

```python
a = np.random.randint(1,10, size=(3,4))

print(a)
print("'''''''''''''''''''")

transposeA = a.T
print(transposeA)
```

Array split

```python
a = np.random.randint(1,10, size=(3,4))

print(a)
print("'''''''''''''''''''")
b = np.random.randint(1,10, size=(7,))

print(b)
print("'''''''''''''''''''")

splitted_arrayA = np.array_split(a,3)
print(splitted_arrayA)
print("'''''''''''''''''''")
splitted_arrayB = np.array_split(b,3)
print(splitted_arrayB)
```
