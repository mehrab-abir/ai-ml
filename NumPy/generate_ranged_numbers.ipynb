```python
import numpy as np

# np.arange(start,end, step)
arr = np.arange(11,20,1)

print(arr)
```

```python
import numpy as np

# np.arange(start,end, step)
arr = np.arange(11,20,2) # step 2

print(arr)
```

reshape the ranged ndarray using *reshape()*

```python
# reshape(shape of the matrix)

import numpy as np

arr = np.arange(1,13,1).reshape(3,4) # 3 rows, 4 colums, [step = 1]

print(arr)
```

### np.logsapce()

```python
import numpy as np

# np.logspace(start,end(included), koyta)
# Default ==> 10^(numbers from the range provided)
arr = np.logspace(1,3,3) # 1 to 3 , 3 ta--10^1, 10^2, 10^3

print(arr)
```

```python
import numpy as np

# np.logspace(start,end(included), koyta)
# Default ==> 10^(numbers from the range provided)
arr = np.logspace(0,3,3) # 0 to 3 , 3 ta--10^0, 10^(something), 10^3

print(arr)
```

```python
import numpy as np

# np.logspace(start,end(included), koyta)
# Default ==> 10^(numbers from the range provided)
arr = np.logspace(1,4,5) # 1 to 4 , 5 ta--10^1, 10^(something from the range)....,10^4

print(arr)
```

With custom base, instead of default 10

```python
import numpy as np

# np.logspace(start,end(included), koyta)
# Default ==> 10^(numbers from the range provided)
arr = np.logspace(1,4,5, base=2) # 1 to 4 , 5 ta and base 2 -->> 2^1, 2^(something from the range)....,2^4

print(arr)
```

```python
import numpy as np

# np.logspace(start,end(included), koyta)
# Default ==> 10^(numbers from the range provided)
arr = np.logspace(0,10,5, base=2) # 5 to 10 , 5 ta and base 2 -->> 2^5, 2^(something from the range)....,2^10

print(arr)
```
