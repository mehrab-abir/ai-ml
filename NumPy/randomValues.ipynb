### *Generate a 1D array with random values* 

```python
import numpy as np

# np.random.rand(size,)
arr = np.random.rand(3,) # 1D ndarray --> 3 items

print(arr)
```

### *Generate a 2D array with random values* 

```python
import numpy as np

# np.random.rand(shape)
arr = np.random.rand(3,4) # 2D --> 3 rows, 4 columns for each row
#values range from 0 to 1

print(arr)
```

```python
import numpy as np
arr = np.random.rand(2,3) # 2D array- 2 rows and 3 colums of random values (0 to 1)
print(arr) 
```

*Make a 2D array with int numbers*

```python
import numpy as np

# np.random.randint(start, end,(shape in a tuple))
arr = np.random.randint(1,10,(3,4)) # 2D --> 3 rows, 4 columns for each row
# values range from 1 to 9 (not 10)

print(arr)
```

```python
import numpy as np

# np.random.randint(start, end,(shape in a tuple))
arr = np.random.randint(1,100,(6,4)) # 2D --> 3 rows, 4 columns for each row
# values range from 1 to 99 (not 100)

print(arr)
```

```python
import numpy as np

# np.random.randint(start, end,(shape in a tuple))
arr = np.random.randint(1,50,(3,2,4)) # 2D --> 3 rows, 4 columns for each row
# values range from 1 to 99 (not 100)

print(arr)
```

*Float numbers*

```python
import numpy as np

# np.random.uniform(start, end,(shape in a tuple))
arr = np.random.uniform(50,100,(4,3)) # 2D --> 3 rows, 4 columns for each row
# values range from 50 to 99 (not 10)

print(arr)
```
