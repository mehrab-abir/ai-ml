```python
import numpy as np
a = np.array([5,3,1,2,6,9])

index = np.where(a==2)

print(index) # (array([3]),)
print(index[0]) # [3]

if(np.where(a==2)[0].size > 0):
    print("Found")
else:
    print("Not found")
```

```python
a = np.array([5,3,1,2,6,9])

index = np.where(a==2)

if(index[0].size > 0):
    print(f"Found at position {index[0]}")
else:
    print("Not found")
```

```python
a = np.array([5,3,1,2,6,9])

index = np.where(a==3)

if(index[0].size > 0):
    print(f"Found at position {index[0][0]}")
else:
    print("Not found")
```

```python
a = np.array([5,3,1,2,6,9])

evens = np.where(a % 2 == 0); # returns index of matched items

print(evens)
```

*np.where(condition,x,y) -- if condition matches, apply x, otherwise y*

```python
a = np.array([5,3,1,2,6,9])

odd_to_zero = np.where(a % 2 == 0,a,0)
print(odd_to_zero)
```

```python
a = np.array([5,3,1,2,6,9])

p = np.where(a < 4,a,0) # if the value is less than 4, keep it as it is, otherwise replace with  0
print(p)
```

**Maximum, minimum -- index and value**

```python
a = np.array([5,3,1,2,6,9])

max_val_idx = np.argmax(a)
min_val_idx = np.argmin(a)

print(max_val_idx)
print(min_val_idx)

print(f"Maximum value : {a[np.argmax(a)]}")
print(f"Minimum value: {a[np.argmin(a)]}")
```
