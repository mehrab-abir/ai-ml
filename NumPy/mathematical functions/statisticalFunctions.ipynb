mean of an array

```python
import numpy as np

arr = np.array([6,4])

print(np.mean(arr))
```

mean of a 2D array (matrix)

```python
matrix = np.random.randint(1,20,(3,4))
print(matrix)

# row wise mean

meanRow = np.mean(matrix, axis = 1)
print(meanRow)

# column wise mean
meanCol = np.mean(matrix, axis = 0)
print(meanCol)
```
