**Log**

```python
import numpy as np
print(np.log10(100)) # 10 base log mane hocche 100 ke kotobar 10 diye vag kora jabe
print(np.log10(1000)) # 10 base log mane hochce 1000 ke kotobar 10 diye vag kora jabe, 1 na asha porjonto --> for 1000, 10^3..answer = 3, jehetu base 10
```

```python
import numpy as np
print(np.log2(8)) # 2 base log mane hocche 8 ke kotobar 2 diye vag kora jabe, 1 na asha porjonto -->> 2^3
print(np.log2(32)) # --> 2^5--answer 5, jehetu base 2
```
