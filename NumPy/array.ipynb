### **ndArray**

```python
import numpy as np

# 1D array
arr1 = np.array([1,2,3,4,5])

# 2D array
arr2 = np.array([
            [1,2,3],
            [5,6,2],
            [7,8,6],
            [6,3,1]
        ])

print(arr1)
print(f"Dimension of arr1 = {arr1.ndim}")
print(f"Shape of the arr1 (1D): {arr1.shape}")
print(f"Size of the arr1 = ", arr1.size) # total number of elements
print(f"Data type: {arr1.dtype}") # type of the data (data must be homogenous in numpy array)
print("---------------------")
print(arr2)
print(f"Dimension of arr2 = {arr2.ndim}")
print(f"Shape of the arr2 (2D): {arr2.shape}")
print(f"Size of the arr2 = ", arr2.size) # total number of elements
print(f"Data type: {arr2.dtype}") # type of the data (data must be homogenous in numpy array)
print("---------------------")
```

```python
import numpy as np

arr2 = np.array([
            [1,2,3],
            [5,6,2],
            [7,8,6],
            [6,3,1]
        ])

# just matrix
for row in arr2:
    for col in row:
        print(col,end=" ")
    print()
```

### *In a 2D NumPy array:*

- Every row must have the same number of columns
- The structure must be rectangular (not jagged)

For 'shape', returns a tuple, first value is how many rows there are, 2nd value is the number of items(columns) each row has.
However, for 1D array, there is only one value and it is just the size of the array

### **3D Array**

**Characteristics:**
*In a NumPy 3D array, every row must have the same number of columns (and every matrix must have the same shape)*

- Every matrix has the same number of rows
- Every row has the same number of columns

```python
# 3D array
arr3 = np.array([
    [
        [5,3,6,1],[5,3,1,9],[7,3,9,8]
    ],
    [
        [4,2,5,1],[3,6,4,5],[3,1,6,4]
    ],
    # [
    #     [6,2,3,1],[3,7,6,5],[1,5,9,4]
    # ]
])

# print(arr3)
print(f"Dimension of arr3 = {arr3.ndim}")
print(f"Shape of the arr3 (3D): {arr3.shape}")
print(f"Size of the arr3 = ", arr3.size) # total number of elements
print(f"Data type: {arr2.dtype}") # type of the data (data must be homogenous in numpy array)
```

*-About the Shape of a 3D Array-*

There are 3 values in the returned tuple. 
1. Number 2D matrix
2. Number rows in each matrix
3. Number columns in each row

**Iterating a 3D array**

```python
arr_3d = np.array([
    [
        [5,3,6,1],[5,3,1,9],[7,3,9,8]
    ],
    [
        [4,2,5,1],[3,6,4,5],[3,1,6,4]
    ],
    # [
    #     [6,2,3,1],[3,7,6,5],[1,5,9,4]
    # ]
])

for matrix in arr_3d:
    for row in matrix:
        for value in row:
            print(value,end=" ")
        print()
    print()
```
