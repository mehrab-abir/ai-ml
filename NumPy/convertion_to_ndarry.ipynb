*From List to ndarray*

```python
import numpy as np

list_a = [6,4,7,8,9]

print(f"Type of list_a: {type(list_a)}")
print(f"List --> {list_a}")

print("----------------")

arr = np.array(list_a)
print(f"Type of the arr: {type(arr)}")
print(f"The ndArray --> {arr}")
```

*From Set to ndarray*

```python
import numpy as np

set_a = {2,3,6,4}
print(f"Type of set_a: {type(set_a)}")
print(f"Set --> {set_a}")

print("----------------")

# arr = np.array(set_a) # returns the same set {}
arr = np.array(list(set_a)) # returns ndarry
print(f"Type of the arr: {type(arr)}")
print(f"The ndArray --> {arr}")
```

*From Dictionary to ndarray*

```python
import numpy as np

dict_a = {
    'a' : 10,
    'b' : 20,
    'c' : 30
}
print(f"Type of dict_a: {type(dict_a)}")
print(f"Dict --> {dict_a}")

print("----------------")

keys = dict_a.keys()
print(keys)

keys_arr = np.array(list(keys))
print(f"Keys to ndarray: {keys_arr}")

print("...................")

values = dict_a.values();
print(values)

values_arr = np.array(list(values))
print(f"Values to ndarray: {values_arr}")
```
