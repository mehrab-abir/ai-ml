```python
import numpy as np

arr = np.random.randint(1,100,(100))

print(arr)
print("=======================")

greather_than_50 = np.count_nonzero(arr > 50)
print(greather_than_50)
```

```python
unique_val = np.unique(arr)
print(unique_val)
```

```python
unique_val = np.unique(arr, return_counts=True)
print(unique_val)
```

```python
unique_val, count = np.unique(arr, return_counts=True)
print(unique_val)
print(count)
```
